```python
# reading time 
with open('list.txt', 'r') as f:
    count = 5

    print(f.read(count), end='')
    print('\n', f.tell())  # tell() returns the current file position (the index of the next character to be read)

    print(f.seek(12))  # seek(0) moves the file pointer back to the beginning of the file/ we can start from anywhere eg start from 10 char

    print(f.read(count))  # reads the next 5 characters from the current file position
```

#Serialization and Deserialization 

Serialization: process of converiting python data type into  json format
Deserialization: process of converiting JSON to python data types

```python
#Serialization using json model:
#list
import json
# list1=[1,2,3,4]
# with open('demo.json','w') as f:
#     json.dump(list1,f)

#dict
dict1={
    'name':'ram',
    'age':20,
    'address':'ktm'
}
with open('demo.json','w') as f:
    json.dump(dict1,f,indent=4)

    
```

```python
#Deserilization
import json 
with open('demo.json','r') as f:
    d=json.load(f)
    print(d)
    print(type(d))
```

```python
#Serialization and Deserialization  in tuple
import json

t=(1,2,3,4,5)

with open('tuple.json','w') as f:
    json.dump(t,f) #provide list
```

```python
with open('tuple.json','r') as f:
    d=json.load(f)
    print(d)
    print(type(d))
```
